```python
# largest of three numbers
def largest (a,b,c):
    if a>=b and a>=c:
        return a
    elif b >=a and b>=c:
        return b
    else:
        return c

a=int(input("Enter a:"))
b=int(input("Enter b:"))
c=int(input("Enter c:"))


print("Largest =",largest(a,b,c))


```

```python
#grade caiculation
def find_grade(marks):
   if marks >=90:
       return "A"
   elif marks >=80: 
       return"B"
   elif marks>=70:
       return"C"
   elif marks>=60:
       return"D"
   elif marks>=50:
       return"E"
   else:
       return"Fail"
marks = int(input("Enter marks:"))
print("Grade=",find_grade(marks))
```
